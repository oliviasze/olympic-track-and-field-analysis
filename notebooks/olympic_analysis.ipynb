```python
# Define field events to exclude
field_events = ['High Jump', 'Long Jump', 'Triple Jump', 'Pole Vault',
                'Shot Put', 'Discus', 'Hammer', 'Javelin',
                'Decathlon', 'Heptathlon']

# Filter to gold medals and track events only
gold = table[table.medal == 'G']
track = gold[~gold.event.str.contains('|'.join(field_events), case=False)]

def calc_improvement(group):
    group = group.sort_values('year')
    first = group.iloc[0]
    last = group.iloc[-1]

    pct_change = ((first['sec'] - last['sec']) / first['sec']) * 100

    return pd.Series({
        'first_year': first['year'],
        'first_result': round(first['sec'], 2),
        'last_year': last['year'],
        'last_result': round(last['sec'], 2),
        'pct_improvement': round(pct_change, 2)
    })

improvement = (track.groupby('event')
                    .apply(calc_improvement)
                    .reset_index()
                    .sort_values('pct_improvement', ascending=False))

print(improvement.to_string(index=False))
```
